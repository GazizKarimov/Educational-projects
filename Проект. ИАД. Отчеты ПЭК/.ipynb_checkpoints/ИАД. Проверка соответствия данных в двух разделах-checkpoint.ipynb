# Поиск несоответствия в разделах 4.2 и 4.3 отчета по ПЭК

**Контекст:** Ежегодно до 25 марта природопользователям, эксплуатирующим объекты ОНВ I,II,III категорий НВОС, необходимо сдавать отчет о проведении производственного экологического контроля, согласно требованиями ФЗ № 7 "Об охране окружающей среды" и 109 Приказа Минприроды РФ.

По разным причинам указанная в отчете информация может содержать ошибки, в частности, расхождения по количеству принятых и переданных отходов между разделами, что является нарушением.

## Цель:
Определить количество отчетов ПЭК с несоответствием между разделами 4.2 и 4.3 (зерно сравнения: ИНН и идентификатор отхода)

## Задачи
1. Проанализировать предоставленные исходные данные;
2. Определить общее количество отчетов ПЭК;
3. Определить количество отчетов ПЭК с нарушением (Риск 7)
4. Сделать выводы

## Содержание

### Определим число отчетов ПЭК

Согласно приказу Минприроды России от 18.02.2022 N 109 отчеты ПЭК сдаются по каждому объекту ОНВ, то есть юр.лицо может сдавать несколько отчетов, если в собственности несколько промплощадок.

Поскольку в предоставленных данных нет информации об объктах ОНВ (номер в реестре объектов ОНВ, факт. адрес), значит, по условиям задачи принимаем, что один ИНН = один отчет ПЭК.

Поскольку считать будем по полю `org_inn`, его стоит привести в единый тип данных int64

```python
pek_4_2['org_inn']=pek_4_2['org_inn'].astype('object')
pek_4_3['org_inn']=pek_4_3['org_inn'].astype('object')
```

```python
pek_4_2['org_inn'].dropna().dtype
```

```python
pek_4_3['org_inn'].dropna().dtype
```

Данные переведены в нужный тип во избежание неточностей в расчетах (например, если ИНН представлен в виде строки)

```python
# Найдем количество уникальных ИНН по таблице 4_2
unique_inn_4_2 = pek_4_2['org_inn'].dropna().nunique()
print(f'Количество отчетов ПЭК по разделу 4_2: {unique_inn_4_2}')
```

```python
# Найдем количество уникальных ИНН по таблице 4_3
unique_inn_4_3 = pek_4_3['org_inn'].dropna().nunique()
print(f'Количество отчетов ПЭК по разделу 4_3: {unique_inn_4_3}')
```

Количество уникальных ИНН в каждом разделе разное, значит, есть предприятия, которые заполняли только одну таблицу

```python
# Проверим пересечение множеств
set_4_2 = set(pek_4_2['org_inn'].dropna())
set_4_3 = set(pek_4_3['org_inn'].dropna())

intersection_count = len(set_4_2 & set_4_3)
print(f'Число общих ИНН по разделам: {intersection_count}')
```

```python
# Найдем число предприятий, которые заполнили только таблицу 4.3
only_4_3 = set_4_3 - set_4_2
count_only_4_3 = len(only_4_3)

# Аналогично, число предприятий, заполнивших только таблицу 4.2
only_4_2 = set_4_2 - set_4_3
count_only_4_2 = len(only_4_2)

print(f'Число предприятий, которые заполнили только таблицу 4.3: {count_only_4_3}, число предприятий, которые заполнили только таблицу 4.2: {count_only_4_2}')
```

Согласно Приказу Минприроды РФ от 15.03.2024 N 173 в таблице 4.2 указывается информация о количестве образованных, переданных, полученных и (или) самостоятельно утилизируемых (или обезвреженных, захороненных и т.д.) отходов. В таблице 4.3 - только информация о переданных или полученных.

Следовательно, мы имеем предприятия, которые ни от кого не получали и никому не передавали отходы (у них заполнена только таблица 4.2), и предприятия, у которых не образовывались отходы, но которые получали и (или) передавали отходы, при этом указав эту информацию только в таблице 4.3.

В первом случае, существование таких предприятий реально, например, если собственные отходы подлежат какому-либо обращению (утилизация, захоронение и т.д.) своими силами. Во втором случае мы наблюдаем явное несоответствие предоставленной в отчете информации.

```python
# Определим общее количество отчетов ПЭК:
all_reports = intersection_count + count_only_4_2 + count_only_4_3

print(f'Общее количество отчетов: {all_reports}')
```

**Итак, общее количество предоставленных отчетов: `58202`, при этом выявлено 179 юрлиц, которые сформировали отчет с нарушениями.**

### Поиск несоответствия разделов ПЭК 4.2 и 4.3 по получению и передаче отходов

**1. Отфильтруем данные так, чтобы оставить только те организации, которые принимали и(или) передавали отходы**

```python
# Посчитаем количество полученных и переданных отходов по каждому ИНН
all_waste_by_inn_4_2 = pek_4_2.groupby(['org_inn'], as_index = False)[['trans_all','waste_received']].sum()

# Отфильтруем таблицу 4.2
trans_or_receive_4_2 = all_waste_by_inn_4_2[(all_waste_by_inn_4_2['trans_all'] > 0) | (all_waste_by_inn_4_2['waste_received'] > 0)].copy()

# Определим количество таких компаний
trans_or_receive_org_4_2 = set(trans_or_receive_4_2['org_inn'])

print(f'Количество компаний, которые получили или передали отходы, согласно таблице 4.2: {len(trans_or_receive_org_4_2)}')
```

**1.1 Для компаний, которые не получали и не передавали отходы, необходимо проверить, есть ли информация о них в разделе 4.3**

```python
# Получим список компаний, которые не передавали и не получали отходы
no_waste_manage_4_2 = all_waste_by_inn_4_2[(all_waste_by_inn_4_2['trans_all'] == 0) & (all_waste_by_inn_4_2['waste_received'] == 0)].copy()

no_waste_manage_org_4_2 = set(no_waste_manage_4_2['org_inn'])

print(f'Количество компаний, которые не получали и не передавали отходы, согласно таблице 4.2: {len(no_waste_manage_org_4_2)}')
```

```python
# Проверим значения:
unique_inn_4_2 == len(trans_or_receive_org_4_2) + len(no_waste_manage_org_4_2)
```

```python
# Проверим вхождение компаний no_waste_manage_org_4_2 в таблицу 4.3
check_risk_7_1 = no_waste_manage_org_4_2 - (no_waste_manage_org_4_2 - set_4_3)

print(f'Количество компаний, указавших обращение с отходами в табл. 4.3, но не указавших это в  табл. 4.2: {len(check_risk_7_1)}')
```

**2. Проверим обратную ситуацию: информация о передаче/получении отходов есть в 4.2, но нет в 4.3**

```python
check_risk_7_2 = trans_or_receive_org_4_2 - set_4_3
print(f'Количество компаний, указавших обращение с отходами в табл. 4.2, но не указавших это в  табл. 4.3: {len(check_risk_7_2)}')
```

**3. Для компаний, которые заполнили оба раздела, необходимо проверить количество полученных и переданных отходов**

```python
# Найдем список компаний, которые заполнили табл 4.2 и табл. 4.3
both_tables_org = trans_or_receive_org_4_2 & set_4_3

len(both_tables_org)
```

Дальнейший анализ соответствий следует проводить не только по ИНН, но и по каждому отходу

```python
# Создадим датафреймы на основе таблиц 4.2 и 4.3,в которых будут только компании, заполнившие оба раздела
df_4_2 = pek_4_2[pek_4_2['org_inn'].isin(both_tables_org)].copy()
df_4_3 = pek_4_3[pek_4_3['org_inn'].isin(both_tables_org)].copy()
```

```python
# Проверим датафрейм на полные дубликаты
dupl_4_2 = df_4_2.duplicated().sum()
dupl_4_3 = df_4_3.duplicated().sum()
print(f'Количество полных дубликатов в таблице 4_2: {dupl_4_2}, в таблице 4.3: {dupl_4_3}')
```

О чем может говорить наличие дубликатов? Отчет ПЭК сдается по каждому объекту ОНВ, следовательно, теоретически, может быть такая ситуация: у юрлица два объекта ОНВ, на каждом объекте образовалось одинаковое количество отхода А, который передается на один и тот же способ обращения (напр. для утилизации), а поскольку в предоставленных файлах отсутствует индентификатор объектов ОНВ, то все строки таблиц по каждой промплощадке сведены под одним ИНН. Отсюда и наличие дубликатов.

Таким образом, удаление дубликатов может привести к потере данных для сравнения.

```python
# Сгруппируем данные по каждому ИНН и отходу для каждого датафрейма
all_by_waste_4_2 = df_4_2.groupby(['org_inn','fkko_code','fkko_name'], as_index = False)[['waste_received','trans_all']].sum()

all_by_waste_4_3 = df_4_3.groupby(['org_inn','fkko_code','fkko_name'], as_index = False)[['receiving_waste','trans_for_processing','trans_for_disposal','trans_for_neutralized','trans_for_storage','trans_for_burial']].sum()

all_by_waste_4_3['trans_all'] = all_by_waste_4_3[['trans_for_processing','trans_for_disposal','trans_for_neutralized','trans_for_storage','trans_for_burial']].sum(axis=1)
```

Предыдущий анализ двух рисков был основан на поиске соответствующих ИНН в обоих разделах 4_2 и 4_3, при этом детализация на уровне конкретных отходов выполнена не была. Следовательно, на данном этапе проверки эта детализация необходима.

Для сравнения строк по категории "ИНН+ФККО" следует объединить таблицы.  Однако стоит учитывать риск того, что какая-либо пара "ИНН+ФККО" из 4.2 не найдет свою в таблице 4_3, или наоборот. В таком случае, весь отчет для конкретного ИНН будет считаться неверным. Чтобы найти такие отчеты, присоединение следует выполнить полное (outer).

```python
# Объединим таблицы и сравним количества полученных и переданных отходов
merge_to_compare = all_by_waste_4_2.merge(all_by_waste_4_3[['org_inn','fkko_code','fkko_name','trans_all','receiving_waste']],
                                          on = ['org_inn','fkko_code','fkko_name'],
                                          how = 'outer',
                                          indicator = True
                                         )

merge_to_compare = merge_to_compare.rename(columns = {'trans_all_x':'trans_all_4_2',
                                                     'waste_received':'waste_received_4_2',
                                                     'trans_all_y':'trans_all_4_3',
                                                     'receiving_waste':'waste_received_4_3'})
```

```python
merge_to_compare.head()
```

Сравнение полей приема и передачи отходов следует провести с учетом еще одного условия. 
При заполнении отчета ПЭК может быть ситуация, когда отход на объекте образовался, но он не передавался и не принимался от сторонних организаций. В этом случае, передача/прием отхода будет равен 0, но сама строка с отходом в таблице 4.2 будет присутствовать, а значит, в 4.3 этого отхода не будет - это не будет являться ошибкой. Следовательно, при поиске ошибочных отчетов следует учитывать только "активные отходы" (которые точно передавались или принимались).

Однако ситуация, когда нулевой отход есть в 4.3, но его нет в 4.2, не имеет смысла, поскольку это бы означало, что отход не образовался, не принимался и не передавался (т.е. строка полностью отсутствует в 4.2), но при этом он передавался/принимался в нулевом количестве - такие отчеты считаются ошибкой.

```python
# 1) Определим отчеты с нарушением: отход есть в 4_2, но нет в 4_3
only_in_4_2 = merge_to_compare[(merge_to_compare['_merge']=='left_only') &
                               ((merge_to_compare['trans_all_4_2'] > 0) | (merge_to_compare['waste_received_4_2'] > 0))].copy()

# 2) Определим отчеты с нарушением: отход есть в 4_3, но нет в 4_2
only_in_4_3 = merge_to_compare[merge_to_compare['_merge']=='right_only'].copy()

# 3) Определим отчеты с нарушением: отход есть в обоих разделах, но объемы не совпадают
waste_mismatch = merge_to_compare[(merge_to_compare['_merge'] == 'both') &
                                  ((merge_to_compare['waste_received_4_2']!=merge_to_compare['waste_received_4_3']) |
                                  (merge_to_compare['trans_all_4_2']!=merge_to_compare['trans_all_4_3']))].copy()

# Посчитаем количество отчетов с данными нарушениями
check_risk_7_3 = set(only_in_4_2['org_inn']) | set(only_in_4_3['org_inn']) | set(waste_mismatch['org_inn'])

print(f'Число компаний, у которых есть расхождения между суммарным количеством принятых/переданных отходов в табл. 4.2 и табл. 4.3: {len(check_risk_7_3)}')
```

**4. Общее количество отчетов ПЭК с ошибками по 7-му риску**

```python
amount_wrong_pek = len(check_risk_7_1) + len(check_risk_7_2) + len(check_risk_7_3)

print(f'Количество отчетов с ошибками по 7-му риску: {amount_wrong_pek}, их процент от общего числа отчетов: {amount_wrong_pek/all_reports:.2%} ')
```

    - ввиду отсутствия информации об объектах ОНВ для каждого ИНН, один уникальный отчет ПЭК соответствует одному уникальному ИНН;
    - в таблице 4.3 указаны ИНН тех же организаций, что и в таблице 4.2, а не ИНН организаций, которым были переданы/от которых приняты отходы, как это прописано в Приказе;
    - в данных теоретически допустимо наличие полных дубликатов.

2. Общее количество уникальных отчетов ПЭК составило 58202 ед.


3. Общее количество отчетов ПЭК с нарушениями по Риску 7 слагается из трех видов ошибок:

| Тип ошибки | Описание | Кол-во отчётов | Доля от отчётов с нарушением | Доля от всех отчётов |
|---|---|---|---|---|
| Риск 7.1 | В разделе 4.3 указан приём/передача отходов, а в разделе 4.2 этой информации нет | 1541 | 10.8% | 2.6% |
| Риск 7.2 | В разделе 4.2 указан приём/передача отходов, а в разделе 4.3 этой информации нет | 1860 | 13.0% | 3.2% |
| Риск 7.3 | Оба раздела заполнены, но не совпадает объём на уровне "ИНН + отход" | 10923 | 76.3% | 18.8% |
| **Итого** | **Отчётов с нарушением по Риску 7** | **14324** | **100%** | **24.61%** |

Итого общее количество отчетов с нарушениями по Риску 7: 14324, или 24,61% от общего числа отчетов.

Данное число является вполне реальным, поскольку отчеты, как правило, заполняются людьми вручную (за исключением специальных софтов от фирмы Экоцентр, где можно в программу загрузить файл информацией из 2-ТП (отходы), 2-ТП (воздух) и автоматически сформировать отчет ПЭК) и при большом количестве наименований отходов часто замыливается глаз и допускаются опечатки/пропуски.
